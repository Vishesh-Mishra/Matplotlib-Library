```python
#color and label
df=pd.read_csv("C:\\Users\\acer\\Documents\\Matplotlib Dataset\\batsman_season_record.csv")
df
```

```python
#multiple bar graphs(little complex)
batsmen = df['batsman']
years = ['2015', '2016', '2017']

# 3. Set up bar positions and width
x = np.arange(len(batsmen))  # base x locations for each batsman
width = 0.25                 # width of each individual bar

fig, ax = plt.subplots(figsize=(10, 6))

# 4. Loop through each year and plot with an offset
multiplier = 0
for year in years:
    offset = width * multiplier
    rects = ax.bar(x + offset, df[year], width, label=year)
    multiplier += 1

# 5. Customize labels, ticks, and legend
ax.set_ylabel('Runs Scored')
ax.set_title('Batsman Performance Across Seasons (2015–2017)')
# Center the x-ticks relative to the grouped bars (using width to align with the middle bar)
ax.set_xticks(x + width, batsmen)
ax.legend(title='Season')

plt.tight_layout()
plt.show()
```

```python
# a problem(in x-axis the names overlap so to overcome this problem )
#xticks
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']
plt.bar(colors,children,color='black')
plt.xticks(rotation="vertical")
```

```python
#Stacked Bar Chart
plt.bar(df["batsman"],df["2017"],label="2017")
plt.bar(df["batsman"],df["2016"],bottom=df["2017"],label="2016")
plt.bar(df["batsman"],df["2015"],bottom=df["2017"]+df["2016"],label="2015")
plt.legend()
plt.show()
```

3D Surface Plots

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
```

```python
xx,yy=np.meshgrid(x,y)
yy.shape
```

```python
z=xx**2+yy**2
z.shape
```

```python
fig=plt.figure()
ax=fig.add_subplot(projection="3d")
p=ax.plot_surface(xx,yy,z,cmap="viridis")
fig.colorbar(p)
```

```python
z=np.sin(xx)+np.cos(yy)
fig=plt.figure()
ax=fig.add_subplot(projection="3d")
p=ax.plot_surface(xx,yy,z,cmap="inferno")
fig.colorbar(p)
```
